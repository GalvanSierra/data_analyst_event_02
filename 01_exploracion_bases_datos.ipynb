# Fase 1 — Exploración de bases de datos
**Objetivo:** comparar 3 bases de datos de al menos 2 tipos distintos y justificar la elección del dataset para el análisis.

## Base 1 — Titanic (tabular, fuente secundaria)
- **Fuente:** competencia "Titanic: Machine Learning from Disaster" de Kaggle (espejo DataScienceDojo). Tipo tabular, fuente **secundaria**.
- **Registros / atributos / tamaño:** 891 filas x 12 columnas (~60 KB en CSV).
- **Documentación:** data dictionary oficial de Kaggle (PassengerId, Survived, Pclass, Name, Sex, Age, SibSp, Parch, Ticket, Fare, Cabin, Embarked). Documentación abundante: cientos de kernels y tutoriales.
- **Aplicaciones:** clasificación binaria (supervivencia), enseñanza de limpieza de datos, imputación, encoding y modelos introductorios.

## Base 2 — MNIST (imagen, fuente secundaria/terciaria)
- **Fuente:** Yann LeCun / NYU (derivada de NIST). Se distribuye como imágenes de 28x28 en escala de grises. Fuente **secundaria** (recopilación curada con propósito de benchmark).
- **Registros / atributos / tamaño:** 70.000 imágenes (60.000 entrenamiento + 10.000 prueba), 784 píxeles por imagen (~50 MB comprimido).
- **Documentación:** página oficial de Yann LeCun con formato de archivos binarios IDX, benchmarks publicados y descripción del preprocesamiento.
- **Aplicaciones:** clasificación de dígitos manuscritos, benchmark de visión por computadora, redes neuronales convolucionales.

## Base 3 — IMDB Reviews (texto, fuente secundaria)
- **Fuente:** Stanford AI Lab (Maas et al., 2011). Reseñas de películas etiquetadas por sentimiento. Fuente **secundaria**.
- **Registros / atributos / tamaño:** 50.000 reseñas (25.000 entrenamiento + 25.000 prueba), texto libre + etiqueta binaria (~80 MB).
- **Documentación:** artículo académico asociado, README del dataset con división train/test balanceada y descripción del etiquetado.
- **Aplicaciones:** análisis de sentimiento, clasificación de texto, embeddings y modelos de lenguaje.
