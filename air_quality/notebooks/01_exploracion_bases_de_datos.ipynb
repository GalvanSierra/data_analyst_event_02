## Fuente
- UCI: https://archive.ics.uci.edu/dataset/360/air+quality
- De Vito et al. (2008), *Sensors and Actuators B*
- Tipo: **secundaria** (recopilada por terceros)

## Características
- 9.471 registros horarios (mar 2004 – feb 2005), 15 atributos, ~785 KB
- Concentraciones reales (`GT`), respuesta de sensores (`PT08.Sx`) y meteorología (`T`, `RH`, `AH`)
- Faltantes codificados como `-200`

## Por qué esta base
Relevante (contaminación urbana real), bien documentada, manejable en Colab, y con variables
continuas muy correlacionadas → ideal para PCA.
